```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

market = pd.read_csv('kino_verified_data .csv', sep=';')
survey = pd.read_csv('user_survey.csv')
history = pd.read_csv('watch_history.csv')

survey = survey.drop_duplicates().dropna().copy()
history = history.drop_duplicates().dropna().copy()

platforms = ['Кинопоиск', 'Okko', 'Иви', 'Wink', 'Остальные']
shares = market.iloc[6:10]['Значение'].astype(float).tolist()
shares.append(100 - sum(shares))

plt.figure(figsize=(8, 5))
plt.pie(shares, labels=platforms, autopct='%1.0f%%')
plt.title('Структура рынка онлайн-кинотеатров РФ (2025)')
plt.show()
```

Вывод:
В России нет одного кинотеатра, где собраны все фильмы: лидер занимает только 31% рынка, а остальной контент разбросан по разным площадкам (Okko, Иви, Wink и др.). Из-за этого зрителю приходится искать фильм сразу в нескольких приложениях. Наш сайт решает эту проблему: он подбирает подходящий фильм в одном месте за пару кликов, чтобы не тратить полчаса на поиски по разным сервисам.

```python

search_times = np.array(survey['avg_search_minutes'])
scores = np.array(survey['would_use_picker_score'])

print('Среднее время поиска фильма, мин:', np.mean(search_times))
print('Медиана времени поиска, мин:', np.median(search_times))
print('Ищут фильм 20 минут и дольше (чел.):', len(search_times[search_times >= 20]))
print('Готовы пользоваться сайтом (оценка 4 и 5, чел.):', len(scores[scores >= 4]))

avg_time_by_score = survey.groupby('would_use_picker_score')['avg_search_minutes'].mean()

plt.bar(avg_time_by_score.index, avg_time_by_score.values)
plt.title('Время поиска фильма у тех, кто готов пользоваться сайтом')
plt.xlabel('Оценка готовности пользоваться сайтом (от 1 до 5)')
plt.ylabel('Среднее время поиска, мин')
plt.show()
```

Вывод:
Большинство опрошенных (оценка 4 и 5) готовы пользоваться сайтом для быстрого подбора фильмов, и именно эти люди тратят на выбор кино по 20–30 минут. График подтверждает прямой спрос.

```python

avail = np.array(history['available_minutes'])
duration = np.array(history['movie_duration_min'])

print('Всего просмотров:', len(history))
print('Фильм оказался длиннее свободного времени:', len(duration[duration > avail]))

company_counts = history['company'].value_counts()

plt.bar(company_counts.index, company_counts.values)
plt.title('С кем зрители смотрят фильмы вечером')
plt.xlabel('Компания')
plt.ylabel('Количество просмотров')
plt.show()
```

Вывод:
Почти половина просмотров проходит в одиночку, треть — с парой, остальные — с друзьями или семьёй. Также во многих случаях фильм оказывается длиннее свободного времени зрителя. Это подтверждает пользу нашего сайта: подбор кино сразу с учётом компании и свободного времени избавляет от долгого поиска.

```python
age_counts = survey['age_group'].value_counts().sort_index()

ages = np.array(survey['age_group'])
print('Зрителей в возрасте 25-34 лет:', len(ages[ages == '25-34']))

plt.plot(age_counts.index, age_counts.values, marker='o')
plt.title('Активность зрителей в зависимости от возраста')
plt.xlabel('Возрастная группа')
plt.ylabel('Количество зрителей')
plt.grid()
plt.show()
```

Вывод:
Активность зрителей достигает пика в возрасте 25–34 лет, а основную аудиторию составляют люди от 16 до 44 лет. Это студенты и работающие молодые люди, у которых вечером мало свободного времени и которые привыкли быстро решать задачи через удобные сайты.
